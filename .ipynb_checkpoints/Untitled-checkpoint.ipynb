```python
# First cell - imports
from ultralytics import YOLO
import os
from pathlib import Path

import pandas as pd
from datetime import datetime
import torch

# Check CUDA availability
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")

# Second cell - Load model and set up paths
model_path = r"E:\Corn_Seed_Analysis\corn11.pt"
test_images_dir = r"E:\Corn_Seed_Analysis\V11_Dataset\test\images"
save_dir = r"E:\Corn_Seed_Analysis\yolo11Results"

# Load model with CUDA
model = YOLO(model_path).to('cuda')

# Create directories
results_dir = Path(save_dir)
results_dir.mkdir(parents=True, exist_ok=True)
(results_dir / "predicted_images").mkdir(exist_ok=True)
(results_dir / "labels").mkdir(exist_ok=True)

# Third cell - Get image list
image_files = [f for f in os.listdir(test_images_dir) 
               if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
print(f"Found {len(image_files)} images to process")

# Fourth cell - Process images
results_list = []

for img_file in image_files:
    try:
        img_path = os.path.join(test_images_dir, img_file)
        
        # Run prediction using CUDA
        results = model(
            source=img_path,
            save=True,
            project=str(results_dir / "predicted_images"),
            name=img_file,
            conf=0.25,  # confidence threshold
            device='cuda'  # explicitly use CUDA
        )
        
        # Process results
        for r in results:
            boxes = r.boxes
            for box in boxes:
                x1, y1, x2, y2 = box.xyxy[0].tolist()
                confidence = box.conf[0].item()
                class_id = box.cls[0].item()
                class_name = model.names[int(class_id)]
                
                results_list.append({
                    'image_file': img_file,
                    'class': class_name,
                    'confidence': round(confidence, 3),
                    'x1': round(x1, 2),
                    'y1': round(y1, 2),
                    'x2': round(x2, 2),
                    'y2': round(y2, 2)
                })
        
        print(f"Successfully processed {img_file}")
        
    except Exception as e:
        print(f"Error processing {img_file}: {str(e)}")
        continue

# Fifth cell - Save results
if results_list:
    df = pd.DataFrame(results_list)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    results_path = results_dir / f"detection_results_{timestamp}.csv"
    df.to_csv(results_path, index=False)
    print(f"Saved detailed results to {results_path}")
    
    summary = df.groupby('class').agg({
        'image_file': 'count',
        'confidence': ['mean', 'min', 'max']
    }).round(3)
    
    summary_path = results_dir / f"detection_summary_{timestamp}.csv"
    summary.to_csv(summary_path)
    print(f"Saved summary to {summary_path}")
    
    print("\nDetection Summary:")
    print(summary)
else:
    print("No detections were made in any of the images")
```
